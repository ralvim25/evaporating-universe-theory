---# Part I: The Action Principle## 1.1 Starting Point: The Scalar Field ActionWe begin with the most general action for a scalar field $\phi$ minimally coupled to gravity:$$S = \int d^4x \sqrt{-g} \left[ \frac{R}{16\pi G} + \mathcal{L}_\phi \right]$$where the scalar field Lagrangian is:$$\mathcal{L}_\phi = -\frac{1}{2}g^{\mu\nu}\partial_\mu\phi\partial_\nu\phi - V(\phi)$$

````
# Symbolic derivationt = sp.Symbol('t', real=True, positive=True)phi = sp.Function('phi')H = sp.Function('H')V = sp.Function('V')Gamma = sp.Symbol('Gamma', positive=True)  # Dissipation ratedisplay(Markdown("### The Scalar Field Lagrangian"))display(Math(r"\mathcal{L}_\phi = \frac{1}{2}\dot{\phi}^2 - V(\phi)"))
````

## 1.2 The Euler-Lagrange EquationVarying the action with respect to $\phi$ in an FRW background gives:$$\ddot{\phi} + 3H\dot{\phi} + \frac{dV}{d\phi} = 0$$This is the **Klein-Gordon equation in an expanding universe**.

````
display(Markdown("### Standard Klein-Gordon (no dissipation)"))display(Math(r"\ddot{\phi} + 3H\dot{\phi} + V'(\phi) = 0"))print("\nThis describes a scalar field rolling down its potential,")print("with Hubble friction (3Hφ̇) from the expanding universe.")
````

---# Part II: The Dissipation Term## 2.1 Adding Friction: The Key InnovationThe Evaporating Universe adds a **dissipation term** to the Klein-Gordon equation:$$\boxed{\ddot{\phi} + 3H\dot{\phi} + V'(\phi) + \Gamma\dot{\phi} = 0}$$where $\Gamma$ is the **dissipation rate** (units: $H_0$).### Physical Interpretation:- The field $\phi$ does not roll freely- It experiences friction against the cosmic medium- Energy lost goes into radiation: $\dot{\rho}_\gamma = \Gamma\dot{\phi}^2$

````
display(Markdown("### The Master Equation"))display(Math(r"\boxed{\ddot{\phi} + (3H + \Gamma)\dot{\phi} + V'(\phi) = 0}"))print("\n" + "="*50)print("THIS IS THE FUNDAMENTAL EQUATION")print("Everything else is derived from this.")print("="*50)
````

## 2.2 Energy Conservation with DissipationThe energy density and pressure of the scalar field are:$$\rho_\phi = \frac{1}{2}\dot{\phi}^2 + V(\phi)$$$$p_\phi = \frac{1}{2}\dot{\phi}^2 - V(\phi)$$With dissipation, energy is transferred to radiation:$$\dot{\rho}_\phi = -3H(\rho_\phi + p_\phi) - \Gamma\dot{\phi}^2$$$$\dot{\rho}_\gamma = -4H\rho_\gamma + \Gamma\dot{\phi}^2$$The term $\Gamma\dot{\phi}^2$ is the **evaporation rate**.

````
display(Markdown("### The Evaporation Rate"))display(Math(r"\boxed{\dot{\rho}_\gamma = \Gamma\dot{\phi}^2}"))print("\nThis is your 'E = mc²':")print("The rate of radiation creation equals Γ times the field velocity squared.")print("\nSymbolically: DE → DM + γ")
````
