## 1.2 Standard $\Lambda$CDM Continuity Equations

In standard cosmology, each component $i$ (baryons, CDM, radiation, dark energy) satisfies
an independent conservation equation:

$$\dot{\rho}_i + 3H(1 + w_i)\rho_i = 0 \tag{1.1}$$

where $H = \dot{a}/a$ is the Hubble parameter and $w_i = p_i/\rho_i$ is the equation-of-state
parameter ($w_c = 0$ for CDM, $w_b = 0$ for baryons, $w_\Lambda = -1$ for $\Lambda$).

This yields the standard scaling:
- Baryons and CDM: $\rho_m \propto a^{-3}$
- Dark energy: $\rho_\Lambda = \mathrm{const.}$

## 1.3 EU Modified Continuity Equations

In the EU framework, the CDM sector couples to the vacuum through an interaction
term $Q(z) = \lambda\,\varepsilon(z)\,H\,\rho_c$, while baryons remain strictly conserved:

$$\dot{\rho}_c + 3H\rho_c = -\lambda\,\varepsilon(z)\,H\,\rho_c \tag{1.2}$$

$$\dot{\rho}_\Lambda = +\lambda\,\varepsilon(z)\,H\,\rho_c \tag{1.3}$$

$$\dot{\rho}_b + 3H\rho_b = 0 \tag{1.4}$$

where:
- $\lambda \in (0,1)$ is the **dissipation fraction** — the fraction of mediator channels
  available for energy transfer (derived in §2),
- $\varepsilon(z)$ is the **redshift-dependent coupling amplitude** (defined below).

**Total energy conservation** is guaranteed by construction:
Eqs. (1.2)+(1.3)+(1.4) sum to $\dot{\rho}_{\rm tot} + 3H(\rho_{\rm tot} + p_{\rm tot}) = 0$.
*The uniqueness of this interaction form is derived from dimensional analysis and general covariance in §1.5 below.*

### 1.3.1 Dimensional Analysis and Covariance of Q

The interaction term $Q$ in Eqs. (1.2)–(1.3) is not postulated — it is the
**unique leading-order form** dictated by FLRW symmetry and decay kinematics.

**Step 1 (Dimension).** From the continuity equation $\dot{\rho} + 3H(\rho + p) = -Q$,
the dimension of $Q$ is $[Q] = [\rho] \times [T^{-1}]$: an energy density times
an inverse time.

**Step 2 (Decay kinematics).** The interaction describes CDM $\to$ vacuum transfer.
By Fermi's golden rule, the transition rate is proportional to the **initial state
population**: $Q \propto \rho_c$. This excludes $Q \propto \rho_\Lambda$ (not a decay),
$Q \propto (\rho_c + \rho_\Lambda)$ (mixes source and sink), and
$Q \propto \rho_b$ or $\rho_r$ (wrong sector).

**Step 3 (Unique clock).** The factor $[T^{-1}]$ must be a homogeneous, isotropic,
first-order geometric scalar. In FLRW, the only candidate is $H = \dot{a}/a$:

- $\dot{H}/H$ is second-order in derivatives (sub-leading);
- $\sqrt{8\pi G\rho/3}$ equals $H$ by Friedmann (not independent);
- Particle physics scales ($m_\nu$, $\Lambda_{\rm QCD}$) break geometric universality;
- Cosmic time $1/t$ is not covariant under time reparametrization.

**Conclusion.** The unique leading-order interaction consistent with general covariance,
FLRW symmetry, and CDM decay kinematics is:

$$Q = (\text{dimensionless}) \times H \times \rho_c \tag{1.5}$$

The dimensionless prefactor decomposes as $\lambda \times \varepsilon(z)$, where
$\lambda = 2/3$ (§2, transversality) and $\varepsilon(z)$ (§6, coupling amplitude)
are each derived from first principles.

**Status: DERIVED** — from FLRW symmetry, decay kinematics, and the leading-order

$$f_{\rm surv}(a) \equiv \exp\!\left[-\lambda\int_0^{\ln a} \varepsilon(a')\,d\ln a'\right] \tag{1.8}$$

$$\frac{d\rho_\Lambda}{d\ln a} = \lambda\,\varepsilon(a)\,\rho_c(a) \tag{1.9}$$

The modified Friedmann equation is therefore:

$$H^2(a) = H_0^2\left[\Omega_b\,a^{-3} + \Omega_c\,a^{-3}\,f_{\rm surv}(a) + \Omega_\Lambda(a)\right] \tag{1.10}$$

where $\Omega_\Lambda(a)$ is obtained by integrating Eq. (1.8) numerically.
The CDM and vacuum terms do **not** cancel algebraically because the
drained CDM energy stops diluting once absorbed by the vacuum.
This is solved numerically by the CLASS Boltzmann code in NB03.

## 2.2 The Landau Gauge and the Transversality Constraint

The GZ construction operates exclusively in the **Landau gauge**, defined by the condition:

$$\partial_\mu A^a_\mu = 0 \tag{2.2}$$

This is a strict constraint: the gauge field must be transverse (divergence-free) at every
spacetime point. In the Landau gauge, the FP operator becomes hermitian
(VZ12, §2.1, Eq. 2.10):

$$-\partial_\mu D_\mu = -D_\mu \partial_\mu \quad \text{(when } \partial_\mu A_\mu = 0\text{)} \tag{2.3}$$

The ghost field $c^a$ satisfies the **ghost equation of motion** derived from the GZ action:

$$\mathcal{M}^{ab}\, c^b = -\partial_\mu D^{ab}_\mu\, c^b = J^a \tag{2.4}$$

where $J^a$ is the source. Within the Gribov region $\Omega$, the positivity condition
$\mathcal{M} > 0$ ensures that the ghost propagator
$\mathcal{G}^{ab}(k) = \langle c^a(k)\,\bar{c}^b(-k)\rangle$ is well-defined.

### 2.2.1 ADM Derivation of Ghost Transversality

The ghost field $c^a_\mu$ is a 4-vector under Lorentz transformations. By the ADM
3+1 foliation (§2.1.1), it decomposes as $c^\mu = (c^0, c^i)$:

- **Temporal component** $c^0$: constrained by the gauge-fixing procedure (analogous
  to the lapse $N$). Not an independent propagating degree of freedom.
- **Spatial components** $c^i$: 3 propagating components on the spatial slice $\Sigma$.

On $\Sigma$, the spatial components admit a scalar-vector-transverse (SVT) decomposition:

$$c^i = c^i_T + \partial^i \varphi \tag{2.5a}$$

where $c^i_T$ satisfies $\partial_i c^i_T = 0$ (2 transverse DOF) and $\varphi$
is the longitudinal scalar (1 DOF).

The Landau gauge $\partial_\mu A^a_\mu = 0$ splits under the 3+1 foliation as
$\partial_0 A^a_0 + \partial_i A^a_i = 0$. The temporal part fixes $c^0$; the
spatial part enforces $\partial_i A^a_i = 0$. Since the FP operator
$\mathcal{M} = -\partial_\mu D_\mu$ acts within the **transverse** functional
space, the longitudinal ghost mode is projected out:

$$\partial_i c^i = \nabla^2 \varphi = 0 \quad \Rightarrow \quad \varphi = 0 \tag{2.5b}$$

(on a compact manifold or with vanishing boundary conditions). Therefore:

$$\boxed{\partial_i c^i = 0} \tag{2.5}$$

**Two independent reasons guarantee this result:**

**A. Gauge-theoretic (Landau gauge).** The GZ framework is defined in Landau gauge
($\partial_\mu A_\mu = 0$), which is not an arbitrary choice but the defining condition
of the Gribov restriction (Gribov 1978, §2). The FP operator $\mathcal{M}$ acts within
the transverse functional space, projecting out the longitudinal mode algebraically:
$\nabla^2 \varphi = 0 \Rightarrow \varphi = 0$ on compact manifolds.

**B. Cosmological (FLRW isotropy).** The ghost field $c^\mu$ is a Grassmann (anticommuting)
variable, so individual ghost VEVs vanish identically: $\langle c^i \rangle = 0$.
However, the Gribov–Zwanziger condensate is a **bilinear** $\langle \bar{c}^a_i c^b_j \rangle$,
and the question is whether its longitudinal component can be non-zero.
A longitudinal bilinear condensate $\langle \bar{c}^a_\parallel c^b_\parallel \rangle \neq 0$
would single out a preferred spatial direction in the vacuum, spontaneously
breaking the isotropy of the FLRW background. Planck observes CMB isotropy to $\delta T/T \sim 10^{-5}$,
constraining any such anisotropic vacuum component to be vanishingly small. Therefore, the
cosmological ground state **requires** the condensate to be purely transverse,
preserving the isometries of the spatial slice $\Sigma$.

Both arguments — one from the internal structure of the GZ framework, the other from the
symmetries of the cosmological background — independently force $\partial_i c^i = 0$.
The transversality condition is not a gauge artifact: it is a physical consequence of the
universe's observed isotropy.

**Status: DERIVED** — from ADM foliation + Landau gauge (gauge-theoretic) AND FLRW
isotropy (cosmological). The longitudinal ghost is killed by both; only the 2 transverse
spatial components $c^i_T$ propagate on $\Sigma$.

## 2.3 Helmholtz Decomposition and Channel Counting

By the **Helmholtz decomposition theorem**, any smooth vector field $\mathbf{v}$ on
$\mathbb{R}^3$ admits a unique orthogonal decomposition into longitudinal (irrotational)
and transverse (solenoidal) components:

$$c^i = c^i_{\parallel} + c^i_{\perp} \tag{2.6}$$

with:
- **Longitudinal**: $c^i_{\parallel} = \partial^i \phi$ for some scalar $\phi$,
  satisfying $\nabla \times \mathbf{c}_{\parallel} = 0$.
- **Transverse**: $\partial_i c^i_{\perp} = 0$ (divergence-free).

The transversality condition (2.5) requires $\partial_i c^i = 0$, which is
automatically satisfied by $c^i_{\perp}$ and imposes
$\partial_i c^i_{\parallel} = \nabla^2 \phi = 0$. For fields that vanish at
spatial infinity, this implies $\phi = 0$, hence $c^i_{\parallel} = 0$.

**Result:** The longitudinal mode is eliminated. Only the two transverse polarizations
survive as dynamical degrees of freedom:

| Ghost component | DOF | Status within $\Omega$ | Equation |
|:----------------|:---:|:-----------------------|:---------|
| Longitudinal $c^i_{\parallel}$ | 1 | **Frozen** (non-dynamical) | $\nabla^2 \phi = 0 \Rightarrow \phi = 0$ |
| Transverse $c^i_{\perp}$ | 2 | **Active** (dynamical) | propagates within $\Omega$ |
| **Total spatial** | **3** | **2 active / 3 total** | |

This counting is identical to the polarization structure of massless gauge bosons in
$d = 4$: a spin-1 field in the Landau gauge carries $d_{\rm spatial} - 1 = 2$ physical
polarizations.

## 2.4 Dissipation Fraction from FLRW Isotropy

The modified continuity equation (Eq. 1.2) contains the interaction term
$Q = \lambda\,\varepsilon(z)\,H\,\rho_c$, which describes the rate of energy transfer
from CDM to the vacuum. This transfer is mediated by the ghost field and can be decomposed
into contributions from each spatial channel:

$$Q = \sum_{i=1}^{3} Q_i = \sum_{i=1}^{3} \Gamma_{ii}\,\varepsilon(z)\,H\,\rho_c \tag{2.7}$$

where $\Gamma_{ij}$ is the **dissipation tensor** encoding the directional structure of
the energy transfer.

The FLRW metric is spatially isotropic: all three spatial directions are physically
equivalent. Under this symmetry, the dissipation tensor must be proportional to the
identity:

$$\Gamma_{ij} = \gamma_0\,\delta_{ij} \tag{2.8}$$

for some scalar $\gamma_0$. This means each spatial channel contributes **equally**
to the dissipation rate when active.

# §3. Derivation of $b = 19/36$: Transverse-Traceless Projection of the Ghost-Graviton Vertex

## 3.1 Physical Motivation

The exponent $b$ in the logistic activation function (Eq. 1.6) controls the sharpness of
the transition at $z_{\rm trans}$. Physically, $b$ encodes the **fraction of the ghost-graviton
interaction that couples to the physical (transverse-traceless) graviton modes** in $d = 4$
spacetime dimensions.

In perturbative quantum gravity, the graviton propagator couples to the Faddeev–Popov ghost
through a cubic vertex $V^\mu_{\;\nu;\rho\sigma}(p, q)$, where $\mu, \nu$ are ghost indices
and $\rho, \sigma$ are graviton (symmetric tensor) indices. Not all components of this vertex
contribute to physically observable processes: the Gribov restriction projects the graviton
onto its **transverse-traceless (TT) sector** — the 2 physical polarizations out of the
$d(d+1)/2 = 10$ symmetric tensor components.

## 3.2 The Ghost-Graviton Vertex

The ghost-graviton interaction vertex in linearized quantum gravity is
(Codello, Percacci & Rahmede 2009, arXiv: 0812.0785):

$$V^\mu_{\;\nu;\rho\sigma}(p, q) = \frac{1}{2}\Big[
\delta^\mu_\rho\, q_\sigma\, p_\nu
+ \delta^\mu_\sigma\, q_\rho\, p_\nu
+ \delta_{\nu\rho}\, p_\sigma\, q^\mu
+ \delta_{\nu\sigma}\, p_\rho\, q^\mu$$
$$\quad - \delta_{\rho\sigma}(q^\mu p_\nu + p^\mu q_\nu)
- \delta^\mu_\nu(p_\rho q_\sigma + p_\sigma q_\rho - \delta_{\rho\sigma}\, p{\cdot}q)
\Big] \tag{3.1}$$

where $p$ is the ghost momentum, $q = p + k$ is the anti-ghost momentum, and
$k$ is the graviton momentum.

## 3.3 The Transverse-Traceless Projector

The TT projector on symmetric rank-2 tensors is the **York decomposition** projector
(York 1973). For a graviton with momentum $k$, it reads:

$$\Pi^{TT}_{\rho\sigma,\lambda\tau}(k) = \frac{1}{2}\left(
P_{\rho\lambda}P_{\sigma\tau} + P_{\rho\tau}P_{\sigma\lambda}
\right) - \frac{1}{d-1}P_{\rho\sigma}P_{\lambda\tau} \tag{3.2}$$

where $P_{\mu\nu}(k) = \delta_{\mu\nu} - k_\mu k_\nu / k^2$ is the transverse projector.
This projector satisfies:
- **Transversality**: $k^\rho \Pi^{TT}_{\rho\sigma,\lambda\tau} = 0$
- **Tracelessness**: $\delta^{\rho\sigma} \Pi^{TT}_{\rho\sigma,\lambda\tau} = 0$
- **Idempotency**: $\Pi^{TT} \cdot \Pi^{TT} = \Pi^{TT}$
- **Rank**: $\mathrm{Tr}(\Pi^{TT}) = d(d-3)/2 + 1 = 2$ for $d = 4$ (the 2 physical graviton polarizations)

The identity on symmetric tensors is:

$$\mathbb{I}_{\rho\sigma,\lambda\tau} = \frac{1}{2}\left(
\delta_{\rho\lambda}\delta_{\sigma\tau} + \delta_{\rho\tau}\delta_{\sigma\lambda}
\right) \tag{3.3}$$

## 3.4 The TT Fraction: Exact Calculation

The exponent $b$ is defined as the **TT fraction of the ghost-graviton vertex**:

$$b \equiv \frac{\mathrm{Tr}\left(V \cdot \Pi^{TT} \cdot V\right)}{\mathrm{Tr}\left(V \cdot \mathbb{I} \cdot V\right)} \tag{3.4}$$

where the traces are taken over all Lorentz indices:

$$\mathrm{Tr}(V \cdot \Pi \cdot V) = \sum_{\mu,\alpha,\rho,\sigma,\lambda,\tau}
V^\mu_{\;\alpha;\rho\sigma}(p,q)\;\Pi_{\rho\sigma,\lambda\tau}(k)\;
V^\alpha_{\;\mu;\lambda\tau}(q,p) \tag{3.5}$$

This is a purely algebraic calculation in $d = 4$ dimensions, performed using
**exact rational arithmetic** (Python `Fraction` class, zero floating-point error).

The computation was verified across 4 independent momentum configurations,

## 4.1 Order Parameter, RG Evolution, and Self-Energy Suppression

At 1-loop, the $\beta$-function receives contributions from the ghost self-energy

## 5.2 The Heat-Kernel Prefactor

The heat-kernel expansion is a fundamental tool in quantum field theory for computing
one-loop effective actions. For an elliptic operator $D$ on a $d$-dimensional manifold $M$,
the trace of the heat kernel admits the asymptotic expansion
(Seeley 1967; DeWitt 1967; Vassilevich 2003):

$$\mathrm{Tr}_{L^2}\!\left(f\,e^{-tD}\right) \cong \sum_{k \geq 0} t^{(k-d)/2}\,a_k(f, D) \tag{5.1}$$

The heat-kernel coefficients $a_k$ are related to the local spectral coefficients $b_k$
by the **universal prefactor** (Vassilevich 2003, Eq. 2.21):

$$a_k(f, D) = (4\pi)^{-d/2} \int_M b_k(x,x)\,f(x)\,d^d x \tag{5.2}$$

The factor $(4\pi)^{-d/2}$ is **universal**: it appears in every one-loop calculation
regardless of the specific operator, field content, or background geometry. It is the
standard phase-space suppression of a single quantum loop.

In $d = 4$ spacetime dimensions:

$$(4\pi)^{-d/2}\Big|_{d=4} = (4\pi)^{-2} = \frac{1}{16\pi^2} \approx 0.006332 \tag{5.2b}$$

This is the same factor that appears in dimensional regularization
($\overline{\rm MS}$ scheme), the Coleman–Weinberg potential, the ABJ anomaly,
and all standard one-loop QFT calculations.

**Proof that the leading coefficient is unity ($c = a_0 = 1$)**:
For a scalar ghost field on a flat FLRW background, the relevant
operator is $D = -\nabla^2$ (Laplacian). The zeroth Seeley–DeWitt
coefficient is $b_0(x,x) = \mathrm{tr}(\mathbb{I}) = 1$ per field
component (Vassilevich 2003, §5.1). Since the ghost is a complex
scalar in the adjoint representation, each color component contributes
exactly $b_0 = 1$, giving $a_0 = (4\pi)^{-d/2} \times 1 = 1/(16\pi^2)$
with **no additional combinatorial factors**.

### 5.2.1 Proof: the factor $(4\pi)^{-d/2}$ is a calculable integral

The universal loop factor is not an assumption — it is the result of a standard
Gaussian integral in $d$-dimensional momentum space:

$$\int \frac{d^dk}{(2\pi)^d}\, e^{-k^2}
= \frac{S_d}{(2\pi)^d}\!\int_0^\infty \!k^{d-1}\,e^{-k^2}\,dk
= \frac{2\pi^{d/2}}{\Gamma(d/2)}\cdot\frac{\Gamma(d/2)}{2\,(2\pi)^d}
= \frac{1}{(4\pi)^{d/2}} \tag{5.2a}$$

where $S_d = 2\pi^{d/2}/\Gamma(d/2)$ is the surface area of the unit sphere in $d$ dimensions.
This identity is verified explicitly in the code cell below for $d = 2, 3, 4, 5, 6$.
The same factor appears in the Coleman–Weinberg effective potential
(Coleman \& Weinberg 1973, Eq. 3.10) as $1/(64\pi^2) = 1/(4 \times 16\pi^2)$,

## 5.3 The UV Cutoff: CKN Bound for the Hubble Sphere

$$\frac{4\pi}{3}\,R_H^3\;\Lambda^4_{\rm UV}

$$\boxed{\Lambda^4_{\rm UV} \leq \frac{3\,M_P^2\,H^2}{8\pi}

The relevant integrals are performed in the **2D transverse momentum plane**.
The use of a 2D measure ($d^2k_\perp$) rather than the 3D volume measure ($d^3k$)

Vandersickel \& Zwanziger 2012, §3.3): in the thermodynamic limit, the functional

$$\boxed{\frac{I_{\rm IR}}{I_{\rm UV}} = \frac{\pi}{4}} \tag{6.4d}$$
