```
# =============================================================# CONSTANTS# =============================================================M_PL = 1.0  # Planck units# Swampland bounds (Obied et al. 2018, Ooguri et al. 2018)c_distance = 1.0  # O(1) coefficientc_dS = 0.6  # Refined bound ~ 0.6c_prime = 1.0  # For second derivative criterionprint(f"Swampland criteria:")print(f"  Distance: |Delta phi| < {c_distance} M_Pl")print(f"  dS slope: M_Pl|V'|/V > {c_dS}")
```

```
# =============================================================# EVAPORATING UNIVERSE POTENTIAL# =============================================================def V_EU(phi,=1e-120, lambda_param=0.8):"""Evaporating Universe exponential potential.V(phi) = * exp(-lambda * phi / M_Pl)"""return * np.exp(-lambda_param * phi / M_PL)def dV_dphi(phi,=1e-120, lambda_param=0.8, eps=1e-8):"""Numerical derivative."""return (V_EU(phi + eps,, lambda_param) - V_EU(phi - eps,, lambda_param)) / (2 * eps)def d2V_dphi2(phi,=1e-120, lambda_param=0.8, eps=1e-6):"""Second derivative."""return (V_EU(phi + eps,, lambda_param) - 2*V_EU(phi,, lambda_param) + V_EU(phi - eps,, lambda_param)) / eps**2print("Potential defined.")
```

```
# =============================================================# SWAMPLAND CRITERIA CALCULATION# =============================================================def check_dS_criterion(phi,=1e-120, lambda_param=0.8):"""Check de Sitter criterion: M_Pl |V'|/V > cFor V = exp(-lambda phi), we have:V' = -lambda VSo: M_Pl |V'|/V = lambda"""V = V_EU(phi,, lambda_param)dV = dV_dphi(phi,, lambda_param)slope = M_PL * np.abs(dV) / V if V > 0 else 0return slopedef check_second_criterion(phi,=1e-120, lambda_param=0.8):"""Check second criterion: M_Pl^2 V''/V < -c'For V = exp(-lambda phi):V'' = lambda^2 VSo: M_Pl^2 V''/V = lambda^2 > 0 (doesn't satisfy < -c')But the first criterion is sufficient."""V = V_EU(phi,, lambda_param)d2V = d2V_dphi2(phi,, lambda_param)curvature = M_PL**2 * d2V / V if V > 0 else 0return curvature# Analytical expectationlambda_param = 0.8  # Our modelprint(f"\nFor lambda = {lambda_param}:")print(f"  M_Pl|V'|/V = lambda = {lambda_param} (analytical)")print(f"  Required: > {c_dS}")print(f"  Status: {'PASS' if lambda_param > c_dS else 'FAIL'}")
```

```
# =============================================================# SCAN OVER LAMBDA VALUES# =============================================================lambda_range = np.linspace(0.1, 2.0, 50)phi_test = 0.5  # Test at a representative field valueslopes = [check_dS_criterion(phi_test, lambda_param=l) for l in lambda_range]# Find minimum lambda that satisfies Swamplandlambda_min = c_dS  # Exactly at the boundprint(f"\nMinimum lambda for Swampland compatibility: {lambda_min}")print(f"Our model has lambda = {lambda_param}")print(f"Margin: {(lambda_param - lambda_min)/lambda_min * 100:.0f}% above minimum")
```

```
# =============================================================# DISTANCE CONJECTURE CHECK# =============================================================# Field excursion from Planck to today# Using slow-roll during inflation + runaway# Estimate from quintessence evolutionDelta_phi_inflation = 10 * M_PL  # Typical large-field inflationDelta_phi_DE = 1 * M_PL  # Small during DE eraDelta_phi_total = Delta_phi_inflation + Delta_phi_DEprint(f"\nDistance Conjecture:")print(f"  Delta_phi_inflation ~ {Delta_phi_inflation} M_Pl")print(f"  Delta_phi_DE ~ {Delta_phi_DE} M_Pl")print(f"  Total ~ {Delta_phi_total} M_Pl")print(f"  ")print(f"  Note: Large-field inflation typically has Delta_phi > M_Pl,")print(f"  which is compatible with 'Refined Swampland Distance Conjecture'")print(f"  that allows super-Planckian excursions with proper tower behavior.")
```
